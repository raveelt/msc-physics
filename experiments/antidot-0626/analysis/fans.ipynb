```python
fig, ax = plt.subplots(1, 1, figsize=(10,4))

V0 = -708
m = 0.08   # V/T

ymin = 200
ymax = 3000

pt.plot_h5_2d(path+'dat106.h5', 'SET5_2d', gradient_axis=None, ax = ax, title = "topgate = 0V", ylabel = "field (mT)", xlabel = "backgate (mV)", yvals = np.linspace(8000, 0, 500), ylim = (ymin,ymax), xlim = (-4000, 3000))
ax.invert_yaxis()

ax.axvline(x=V0, linestyle='--',color="red", ymin=0, ymax=1, lw=1) #V_CNP
B_line = np.array([ymin, ymax])
nus = np.arange(1,10,1)
for nu in nus:
    ax.plot(V0 + m*B_line*nu, B_line, color='red', linestyle='--', lw = 1)
    ax.plot(V0 - m*B_line*nu, B_line, color='red', linestyle='--', lw = 1)
```
